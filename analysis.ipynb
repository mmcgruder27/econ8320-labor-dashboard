This project uses monthly U.S. labor market data from the Bureau of Labor Statistics Public API.  
The dataset currently covers January 2022 through the most recent available month at the time of data collection.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("labor_data.csv")
df.head()
```

```python
df.info()
df.isna().sum()
```

```python
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date")

df.info()
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df["date"], df["unemployment_rate"])
plt.title("Unemployment Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Percent")
plt.show()
```

The unemployment rate shows a general decline from 2022 to 2023.

```python
plt.figure()
plt.plot(df["date"], df["nonfarm_employment"])
plt.title("Total Nonfarm Employment Over Time")
plt.xlabel("Date")
plt.ylabel("Employment (Thousands)")
plt.show()
```

Total nonfarm employment increases steadily over time, which aligns with the low unemployment rates seen during the same period


```python
plt.figure()
plt.plot(df["date"], df["labor_force_participation"])
plt.title("Labor Force Participation Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Percent")
plt.show()
```

The labor force participation rate remains relatively stable, with minor fluctuations across the observed period.


```python
plt.figure()
plt.plot(df["date"], df["average_hourly_earnings"])
plt.title("Average Hourly Earnings Over Time")
plt.xlabel("Date")
plt.ylabel("Dollars")
plt.show()
```

Average hourly earnings increase steadily over time, indicating ongoing wage growth in the U.S. labor market.
